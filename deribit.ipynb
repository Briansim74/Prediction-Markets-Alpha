```python
import requests

url = "https://www.deribit.com/api/v2/public/get_instruments"

params = {
    "currency": "BTC",
    "kind": "option"
}

response = requests.get(url, params=params)

data = response.json()["result"]

target_expiry = "25DEC26"

options = [
    x for x in data
    if target_expiry in x["instrument_name"]
]

# for option in data[:5]:
#     print(
#         option["instrument_name"],
#         option["strike"],
#         option["expiration_timestamp"]
#     )

url = "https://www.deribit.com/api/v2/public/get_book_summary_by_currency"

params = {
    "currency": "BTC",
    "kind": "option"
}

options = requests.get(
    url,
    params=params
).json()["result"]

for x in options[:3]:
    print(
        x["instrument_name"],
        x["mark_iv"],
        x["underlying_price"]
    )

curve = []

for option in options:

    if option["instrument_name"].endswith("-C"):

        curve.append({
            "strike": option["strike"],
            "iv": option["mark_iv"],
        })
```
